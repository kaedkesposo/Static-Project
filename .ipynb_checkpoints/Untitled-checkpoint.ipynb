```python
# loading and processing 5Essentials data

ai = pd.read_csv('./data/5Essentials/Ambitious Instruction.csv')
ct = pd.read_csv('./data/5Essentials/Collaborative Teachers.csv')
el = pd.read_csv('./data/5Essentials/Effective Leaders.csv')
invf = pd.read_csv('./data/5Essentials/Involved Families.csv')
se = pd.read_csv('./data/5Essentials/Supportive Environment.csv')

def cols_re(prefix, df):
    rename_dict = {col: f"{prefix}: {col}" for col in df.columns[1:]}
    df = df.rename(columns=rename_dict)
    return df

ai = cols_re('AI', ai)
ct = cols_re('CT', ct)
el = cols_re('EL', el)
invf = cols_re('IF', invf)
se = cols_re('SE', se)

fiveE = ai.merge(ct).merge(el).merge(invf).merge(se)
fiveE
```
